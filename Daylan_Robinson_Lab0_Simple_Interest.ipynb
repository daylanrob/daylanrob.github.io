## **Introduction**

### In finance, the **Simple Interest, S.I**, made on any money that is a person borrows from a bank is calculated as:
### P x R x T,
### where:
### **P** = Principle in dollars $,
### **R** = Rate in percent,
### **T** = Time in years.

### We're going to design a program that calculates the **Simple Interest, S.I**. Then you can use it to collect interest from the money you borrow to your friends!





### Run the cell below

```python
# Simple interest calculator

# Run the code below
P = int(input('Enter the principle, P: '))
R = int(input('Enter the Rate, R: '))/100
T = int(input('Enter the time, T in years: '))

# Calculate the simple interest
SI = P*R*T
print('The simple interest for Principle P =',P, 'and Rate R=',R,'Time T = ',T,'is: $',SI)
```
